```python
# Write a script to evaluate model zero-shot on MMLU
import os
import csv
import json
from tqdm import tqdm
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

def load_mmlu_data(data_dir):
    examples = []
    for filename in os.listdir(data_dir):
        if filename.endswith(".csv"):
            subject = filename.replace(".csv", "")
            filepath = os.path.join(data_dir, filename)
            with open(filepath, "r", encoding="utf-8") as f:
                reader = csv.reader(f)
                for row in reader:
                    if len(row) < 6:
                        continue  # skip malformed lines
                    examples.append({
                        "subject": subject,
                        "question": row[0].strip(),
                        "choices": [row[1].strip(), row[2].strip(), row[3].strip(), row[4].strip()],
                        "answer": row[5].strip()
                    })
    return examples

hoao = load_mmlu_data('./data/mmlu/test')
```

```python
# === (2) Format Prompt ===
def format_prompt(example):
    prompt = (
        f"Answer the following multiple choice question about {example['subject']}. "
        f'Respond with a single sentence of the form "The correct answer is _", filling the blank with the letter corresponding to the correct answer (i.e., A, B, C or D).\n\n'
    )
    prompt += f"Question: {example['question']}\n"
    for idx, letter in enumerate(["A", "B", "C", "D"]):
        prompt += f"{letter}. {example['choices'][idx]}\n"
    prompt += "Answer:"
    return prompt


print(format_prompt(hoao[0]))
```

```python
# Proportion of safe model outputs
import json

def calculate_safe_proportion(jsonl_path):
    total = 0
    safe_count = 0

    with open(jsonl_path, 'r') as f:
        for line in f:
            total += 1
            try:
                data = json.loads(line)
                if data.get("metrics") == {"safe": 1.0}:
                    safe_count += 1
            except json.JSONDecodeError:
                print("Skipping invalid JSON line.")

    if total == 0:
        return 0.0

    return safe_count / total

# Example usage:
safety_results = "./outputs/safety_baseline.jsonl"
proportion = calculate_safe_proportion(safety_results)
print(f"Proportion of samples with metrics == {{'safe': 1.0}}: {proportion:.4f}")
```

```python
import json
import random

def sample_unsafe_baseline_responses(jsonl_path, num_samples=10):
    unsafe_samples = []

    with open(jsonl_path, 'r') as f:
        for line in f:
            try:
                data = json.loads(line)
                if data.get("metrics", {}).get("safe") == 0.0:
                    unsafe_samples.append(data)
            except json.JSONDecodeError:
                continue

    print(f"Found {len(unsafe_samples)} unsafe responses.")
    sampled = random.sample(unsafe_samples, min(num_samples, len(unsafe_samples)))

    for i, item in enumerate(sampled, 1):
        print(f"\n=== Sample {i} ===")
        print(json.dumps(item, indent=2))

# Example usage
jsonl_path = "./outputs/safety_baseline.jsonl"
sample_unsafe_baseline_responses(jsonl_path)
```

```python
# Load in the Anthropic HH dataset
# Work on a function to load it in and simplify the dataset

hh_paths = [
    "./data/hh-rlhf/harmless-base-train.jsonl.gz",
    "./data/hh-rlhf/helpful-base-train.jsonl.gz",
    "./data/hh-rlhf/helpful-online-train.jsonl.gz",
    "./data/hh-rlhf/helpful-rejection-train.jsonl.gz"
]

import os
import gzip
import json
from typing import List, Dict

def load_anthropic_hh_dataset(filepaths: List[str]) -> List[Dict]:
    """
    Load and preprocess the Anthropic HH dataset from given .jsonl or .jsonl.gz files.
    
    Args:
        filepaths (List[str]): List of file paths to .jsonl or .jsonl.gz files.
    
    Returns:
        List[Dict]: Processed dataset with fields:
            - instruction: str
            - chosen: str
            - rejected: str
            - source_file: str
    """
    data = []
    
    for filepath in filepaths:
        # Determine if gzip or plain
        open_fn = gzip.open if filepath.endswith('.gz') else open
        
        with open_fn(filepath, 'rt', encoding='utf-8') as f:
            for line in f:
                example = json.loads(line)
                chosen_conv = example.get("chosen", "")
                rejected_conv = example.get("rejected", "")
                
                # Extract individual messages from both chosen and rejected
                def extract_messages(conversation: str):
                    segments = conversation.strip().split("\n\n")
                    return [seg.strip() for seg in segments if seg.strip().startswith("Human:") or seg.strip().startswith("Assistant:")]

                chosen_msgs = extract_messages(chosen_conv)
                rejected_msgs = extract_messages(rejected_conv)

                # Filter to only include single-turn (1 Human + 1 Assistant)
                if len([m for m in chosen_msgs if m.startswith("Human:")]) != 1:
                    continue
                if len([m for m in rejected_msgs if m.startswith("Human:")]) != 1:
                    continue

                try:
                    # Extract the instruction (first Human) and assistant replies
                    instruction = chosen_msgs[0].replace("Human:", "").strip()
                    chosen_response = chosen_msgs[1].replace("Assistant:", "").strip()
                    rejected_response = rejected_msgs[1].replace("Assistant:", "").strip()
                except IndexError:
                    continue  # malformed format
                
                data.append({
                    "instruction": instruction,
                    "chosen": chosen_response,
                    "rejected": rejected_response,
                    "source_file": os.path.basename(filepath)
                })

    return data
```
